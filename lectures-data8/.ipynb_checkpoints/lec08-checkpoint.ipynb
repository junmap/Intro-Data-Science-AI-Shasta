## Height ##

### Question: What is the height of the [40, 65] bin?

```python
top_movies.hist('Age', bins=my_bins, unit='Year')
```

```python
# Step 1: Calculate % of movies in the [40, 65) bin
percent = binned_data.where('bin', 40).column('Percent').item(0)
```

```python
# Step 2: Calculate the width of the 40-65 bin
width = 65 - 40
```

```python
# Step 3: Area of rectangle = height * width
#         --> height = percent / width
height = percent / width
height
```

### What are the heights of the rest of the bins?

```python
# Get the bin lefts
bin_lefts = binned_data.take(np.arange(binned_data.num_rows - 1))
bin_lefts
```

```python
# Get the bin widths
bin_widths = np.diff(binned_data.column('bin'))
bin_lefts = bin_lefts.with_column('Width', bin_widths)
bin_lefts
```

```python
# Get the bin heights
bin_heights = bin_lefts.column('Percent') / bin_widths
bin_lefts = bin_lefts.with_column('Height', bin_heights)
```

```python
bin_lefts
```

# Solution to the discussion question at the end of the Lecture 08 slides

```python
# create income table according to the data on the slides
income = Table().with_column('Income',make_array(61.7,57.5,40,24.75,24,24,20,20,15,15,15,12,10.5,10.5,10,10,8.5,8,6,4.5))
income
```

```python
my_bins = make_array(0,15,25,85)
income.hist('Income', bins = my_bins, unit ='Millions')
```

```python
# calculate bin heights
income_binned = income.bin('Income', bins = my_bins)
income_binned
```

```python
income_percent = 100 * income_binned.column('Income count') / sum(income_binned.column('Income count'))
income_binned = income_binned.with_column('Percent',income_percent)
income_binned
```

```python
income_left = income_binned.take(np.arange(income_binned.num_rows-1))
income_left
```

```python
income_width = np.diff(income_binned.column('bin'))
income_left = income_left.with_columns('Width',income_width)
income_left
```

```python
income_height = income_left.column('Percent') / income_width
income_left = income_left.with_column('Height',income_height)
income_left
```
